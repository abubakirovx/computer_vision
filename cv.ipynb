```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
image = cv2.imread("/content/cat_image.jpg")

cv2_imshow(image)
```

```python
print("SHAPE",image.shape)
# 463 => height
# 703 => width
# 3 => channels (BGR) BLUE GREEN RED
print("HEIGHT",image.shape[0])
print("WIDTH",image.shape[1])
print("CHANNELS",image.shape[2])
```

```python
(b,g,r)=image[100,50]
print("blue",b,"green",g,"red",r)
```

```python
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized=cv2.resize(gray,(400,400))
cv2_imshow(resized)
```

```python
cropped=gray[100:300,150:350]
cv2_imshow(cropped)
```

```python
rotated=cv2.rotate(cropped,cv2.ROTATE_180)
cv2_imshow(rotated)
```

```python
flipped=cv2.flip(gray,1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.jpg",flipped)
```

```python
image2=cv2.imread("/content/flipped_cat.jpg")
cv2_imshow(image2)
```
